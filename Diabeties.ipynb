```python
df2['Glucose'] = df2['Glucose'].replace(0,df2['Glucose'].median())
df2['BloodPressure'] = df2['BloodPressure'].replace(0,df2['BloodPressure'].median())
df2['SkinThickness'] = df2['SkinThickness'].replace(0,df2['SkinThickness'].median())
df2['Insulin'] = df2['Insulin'].replace(0,df2['Insulin'].median())
df2['BMI'] = df2['BMI'].replace(0,df2['BMI'].median())
```

```python
# dividing the data into independent and dependent varaibles

X = df2.drop('Outcome',axis=1)
y = df2['Outcome']
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
scalar = StandardScaler()
```

```python
X_train = scalar.fit_transform(X_train)
X_test = scalar.transform(X_test)
```

```python
models = {'Lr':LogisticRegression(),
          'SVC':SVC(),
          'DT':DecisionTreeClassifier(),
          'Rf':RandomForestClassifier()
          }
```

```python
def evaluvate_model(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    matrix = confusion_matrix(y_true, y_pred)
    report = classification_report(y_true, y_pred)
    acc = accuracy_score(y_true, y_pred)

    return matrix, report, acc
```

```python
model_list = []
acc_score2 = []

for i in range(len(list(models))):
    model=list(models.values())[i]
    model.fit(X_train,y_train)

    #Make Predictions
    y_pred=model.predict(X_test)

    matrix, report, acc =evaluvate_model(y_test.values,y_pred)

    print(list(models.keys())[i])
    model_list.append(list(models.keys())[i])

    print('Model Training Performance')
    print("matrix:",matrix)
    print("report:",report)
    print("acc_score",acc)

    acc_score2.append(acc)
    
    print('='*35)
    print('\n')
```

```python
models = {
    "LogisticRegression": {
        "model": LogisticRegression(max_iter=1000),
        "params": {
            "C": [0.01, 0.1, 1, 10],
            "penalty": ["l2"],
            "solver": ["lbfgs"]
        }
    },

    "RandomForest": {
        "model": RandomForestClassifier(),
        "params": {
            "n_estimators": [100, 200, 300],
            "max_depth": [None, 5, 10, 20],
            "min_samples_split": [2, 5, 10]
        }
    },

    "SVM": {
        "model": SVC(),
        "params": {
            "C": [0.1, 1, 10],
            "kernel": ["linear", "rbf"],
            "gamma": ["scale", "auto"]
        }
    }
}
```

```python
best_models = {}

for model_name, model_info in models.items():
    print(f"Running RandomizedSearchCV for {model_name}")

    rscv = RandomizedSearchCV(
        estimator=model_info["model"],
        param_distributions=model_info["params"],
        n_iter=10,
        cv=5,
        scoring="accuracy",
        random_state=42,
        n_jobs=-1
    )

    rscv.fit(X_train, y_train)

    best_models[model_name] = {
        "best_score": rscv.best_score_,
        "best_params": rscv.best_params_,
        "best_estimator": rscv.best_estimator_
    }

    print("Best Score:", rscv.best_score_)
    print("Best Params:", rscv.best_params_)
    print("best_estimator:", rscv.best_estimator_)
    print("-" * 40)
```
